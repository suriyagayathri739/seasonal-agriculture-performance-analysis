```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("seasonal_agriculture_performance_dataset.csv")
print(df.shape)
df.head()
```

```python
print(df.dtypes)
print(df.isna().sum()[df.isna().sum() > 0])
```

```python
clean = df.copy()
for col in clean.select_dtypes(include="number").columns:
    clean[col] = clean[col].fillna(clean[col].median())
print("Remaining missing values:", clean.isna().sum().sum())
```

```python
season_order=["Kharif","Rabi","Zaid"]
clean["Season"]=pd.Categorical(clean["Season"],categories=season_order,ordered=True)
season_summary=clean.groupby("Season",observed=True).agg(
    Farms=("Farm_ID","count"),
    Avg_Yield=("Yield_Tonnes_Ha","mean"),
    Median_Yield=("Yield_Tonnes_Ha","median"),
    Avg_Profit=("Profit_INR","mean"),
    Median_Profit=("Profit_INR","median"),
    Positive_Profit_Pct=("Profit_INR",lambda x:100*(x>0).mean()),
    Avg_Water_Efficiency=("Water_Efficiency_t_per_1000m3","mean")
).round(2)
season_summary
```

```python
crop_season=clean.pivot_table(index="Crop",columns="Season",values="Yield_Tonnes_Ha",aggfunc="mean",observed=True).round(2)
crop_season
```

```python
irrigation_season=clean.pivot_table(index="Irrigation_Method",columns="Season",values="Profit_INR",aggfunc="mean",observed=True).round(0)
irrigation_season
```

```python
corr=clean.select_dtypes("number").corr()
yield_corr=corr["Yield_Tonnes_Ha"].sort_values(ascending=False)
yield_corr.head(10)
```

## Key findings

- Kharif has the highest average yield and average profit.
- Zaid has the lowest average profit and lowest share of profitable farms.
- Sugarcane has a much higher yield scale than the other crops in this dataset.
- Kharif has the highest average yield for every listed crop.
- Drip has the highest overall average profit among irrigation methods and remains positive on average in all three seasons.
- Water-efficiency is strongly associated with yield (about r = 0.92).
- Rainfall has a weak linear association with yield (about r = 0.03).

## Recommendations

Use seasonal crop/resource comparisons for planning; investigate weak Zaid profitability; evaluate irrigation using both profit and water efficiency; and add multi-year weather and market data for future predictive analysis.

## Limitation

The results are observational patterns in the supplied dataset and do not establish causal relationships.
